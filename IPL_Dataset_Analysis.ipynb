**Most Winning Team**

```python
matches['winner'].value_counts()
```

```python
matches['winner'].value_counts().head(10)
```

**Top winning teams bar chart**

```python
matches['winner'].value_counts().head(10).plot(kind='bar')

plt.title('Top Winning Teams')
plt.xlabel('Teams')
plt.ylabel('Wins')
plt.xticks(rotation=45)
plt.show()
```

**Countplot of winning teams**

```python
plt.figure(figsize=(12,6))

sns.countplot(x='winner', data=matches,
order=matches['winner'].value_counts().index[:10])

plt.title('Top Winning Teams')
plt.xticks(rotation=45)
plt.show()
```

**Toss Winner Analysis**



```python
matches['toss_winner'].value_counts().head()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(x='toss_winner',
data=matches,
order=matches['toss_winner'].value_counts().index[:10])

plt.xticks(rotation=45)
plt.title('Top Toss Winning Teams')
plt.show()
```

**Toss Decision Analysis**

```python
matches['toss_decision'].value_counts()
```

```python
matches['toss_decision'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Toss Decision Distribution')
plt.ylabel('')
plt.show()
```

**Matches Played Per Season**



```python
season_matches = matches['season'].value_counts().sort_index()
print(season_matches)
```

```python
plt.figure(figsize=(10,5))

sns.lineplot(x=season_matches.index,
y=season_matches.values)

plt.title('Matches Played Per Season')
plt.xlabel('Season')
plt.ylabel('Matches')
plt.show()
```

**Venue Analysis**

```python
matches['venue'].value_counts().head(10)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=matches['venue'].value_counts().head(10).values,
    y=matches['venue'].value_counts().head(10).index
)

plt.title('Top Venues')
plt.show()
```

**Team Winning Percentage**

```python
wins = matches['winner'].value_counts()

winning_percentage = (wins / len(matches)) * 100

print(winning_percentage.head())
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=list(winning_percentage.index),
    y=list(winning_percentage.values)
)
plt.title('Team Winning Percentage')
plt.xlabel('Teams')
plt.ylabel('Winning Percentage')
plt.xticks(rotation=45)
for index, value in enumerate(winning_percentage.values):
    plt.text(index, value + 0.2, f'{value:.1f}%')
plt.show()
```

**Correlation Heatmap**

```python
numeric_data = matches.select_dtypes(include=['int64', 'float64'])
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(numeric_data.corr(), annot=True)

plt.title('Correlation Heatmap')
plt.show()
```
